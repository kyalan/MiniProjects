## Scrapping all dramas in Dorama.info

```python
import pandas as pd
from Scarper import Scarper_dorama

years = list(range(2010, 2022))
seasons = list(range(1, 5))
```

```python
df_dorama = pd.DataFrame()
scraper = Scarper_dorama()

for year in years:
    for season in seasons:
        df_season = scraper.crawl(year, season, verbose=True)
        df_dorama = df_dorama.append(df_season, ignore_index=True)
```

```python
df_dorama.to_csv('df_dorama.csv')
```

## Scrapping all dramas in wikipedia

```python
import pandas as pd
from Scarper import Scarper_wiki

years = list(range(2014, 2022))
```

```python
df_wiki = pd.DataFrame()
scraper = Scarper_wiki()

for year in years:
    df_year = scraper.crawl(year, verbose=True)
    df_wiki = df_wiki.append(df_year, ignore_index=True)
```

```python
# df_wiki.to_csv('df_wiki.csv')
df_wiki.to_pickle('df_wiki.pkl')
```

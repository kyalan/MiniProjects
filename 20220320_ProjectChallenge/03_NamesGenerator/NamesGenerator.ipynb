# Importing Database

```python
import pandas as pd
import numpy as np
import random
```

```python
df_actress = pd.read_csv('df_actress.csv')
```

# Generating Names

```python
age_input = 20
```

Calculating Probability Distribution <br>
$$P(\text{actress}) \propto \frac{1}{1+|\text{age_difference}|}$$

```python
probabilities = 1.0 / (1.0 + np.abs(df_actress.age - age_input)**2)
probabilities = probabilities / probabilities.sum()
```

Generating Names

```python
randnums = random.choices(df_actress.index, weights=probabilities, k=2)

family_gen, given_gen = df_actress.family[randnums[0]], df_actress.given[randnums[1]]
names_gen = family_gen + ' ' + given_gen
print(names_gen)
print(f'Surname Info:')
print(df_actress.query(f'family == "{family_gen}"').head())
print(f'Firstname Info:')
print(df_actress.query(f'given == "{given_gen}"').head())
```

## Input Artists and other parameters

```python
artists = ['有村架純', '新垣結衣', '石原さとみ', '菜々緒', '永野芽郁'
           , '吉田羊', '木村文乃', '吉岡里帆', '上野樹里', '綾瀬はるか'
           , '森七菜', '多部未華子', '上白石萌音', '上白石萌歌', '高畑充希'
           , '吉高由里子', '戸田恵梨香', '長澤まさみ', '波瑠'
          ]
npost = 25

ig_player = ky_ig_player()
```

## Artist Dictionary

```python
dict_girl2ig = {artist : ig_player.get_ig_username(artist) for artist in artists}
dict_ig2girl = {v : k for k, v in dict_girl2ig.items()}
```

## Download

```python
for username in dict_ig2girl:
    ig_player.download_ig_posts(username, npost)
    
    if os.path.isdir(username):
        os.rename(username, dict_ig2girl[username])
        print(f'Folder {username} is changed to {dict_ig2girl[username]}.')
    
    # remove non-jpg files
    files = [os.path.join(dict_ig2girl[username], file) for file in os.listdir(dict_ig2girl[username]) if not file.endswith(".jpg")]
    [os.remove(file) for file in files]
```

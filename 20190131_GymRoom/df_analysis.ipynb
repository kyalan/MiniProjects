# df.csv analysis

This notebook loads `df.csv`, inspects its structure, and extracts all facility names from the equipment-related columns.

```python
from pathlib import Path
import re

import pandas as pd
```

```python
csv_path = Path('df.csv')
df = pd.read_csv(csv_path)

print(f'Loaded {csv_path} with shape: {df.shape}')
print('First 20 columns:')
print(df.columns[:20].tolist())
df.head()
```

```python
base_columns = {'venue', 'address', 'tel', 'lat', 'long', 'href', 'id', 'area'}

def clean_facility_name(name: str) -> str:
    cleaned = re.sub(r'\s+', ' ', name).strip()
    cleaned = re.sub(r'\s*\*\s*$', '', cleaned).strip()
    return cleaned

facility_names = sorted({
    clean_facility_name(column.removeprefix('facility_group_').removeprefix('facility_'))
    for column in df.columns
    if column not in base_columns
    and clean_facility_name(column.removeprefix('facility_group_').removeprefix('facility_'))
})

facility_names_df = pd.DataFrame({'facility_name': facility_names})
print(f'Found {len(facility_names_df)} unique facility names')
facility_names_df.head(25)
```

```python
output_path = Path('facility_names_from_df.csv')
facility_names_df.to_csv(output_path, index=False, encoding='utf-8-sig')
print(f'Saved facility names to: {output_path.resolve()}')
```

```python
import os
import shutil
import re
import time
from datetime import datetime as dt
```

```python
dir_str = r'D:\Album\DCIM\WhatsApp Video - test set'
```

```python
mp4list = [os.path.join(dir_str, f_str) for f_str in os.listdir(dir_str) if re.search('mp4$', f_str)]

for filepath in mp4list:  
    msecs = os.path.getmtime(filepath)
    mdt = dt.fromtimestamp(msecs)
    newfilename = mdt.strftime('%Y%m%d_%H%M%S')
    newfilepath = os.path.join(dir_str, newfilename) + '.mp4'

    i = 0
    while (filepath != newfilepath) & (os.path.exists(newfilepath)):
        i += 1
        newfilename_i = newfilename + f'_{i}'
        newfilepath = os.path.join(dir_str, newfilename_i) + '.mp4'

    shutil.move(filepath, newfilepath)
```
